```python
import re
import pandas as pd

months = [
    "JAN","FEB","MAR","APR","MAY","JUN",
    "JUL","AUG","SEP","OCT","NOV","DEC","TOTAL"
]

rows = []

section = None
store = None

with open("C:\Users\USER\kerjaan\LR_TREND_PERTOKO_REG_BDG1_0626.txt", encoding="utf-8") as f:

    for line in f:

        line = line.rstrip()

        # metadata
        if "Toko=" in line:
            store = line.strip()

        # skip
        if (
            line == ""
            or "--------" in line
            or "========" in line
            or "PTD-Actual" in line
            or "Currency" in line
            or "Comparative" in line
            or "Current Period" in line
            or "Date:" in line
        ):
            continue

        # section
        if (
            line.isupper()
            and not re.search(r"\d", line)
            and len(line.split()) <= 6
        ):
            section = line.strip()
            continue

        # ambil angka di belakang
        nums = re.findall(r'-?[\d,]+', line)

        if len(nums) >= 13:

            values = nums[-13:]

            account = line

            for v in values:
                account = account.replace(v, "", 1)

            account = account.strip()

            record = {
                "Store": store,
                "Section": section,
                "Account": account
            }

            for c, v in zip(months, values):
                record[c] = int(v.replace(",", ""))

            rows.append(record)

df = pd.DataFrame(rows)

print(df.head())

df.to_excel("C:\Users\USER\kerjaan\hasil_etl.xlsx", index=False)
```
